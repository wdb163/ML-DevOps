```python
import seaborn # Seaborn is a Python data visualization library based on matplotlib.
import pandas as pd # Python Data Analysis Library
import wandb # Weights and Biases
import ydata_profiling # A highly popular Python library that automates the Exploratory Data Analysis (EDA) process
from IPython.display import IFrame

iframe = IFrame("profiling_report.html", width=1000, height=700)
```

```python
run = wandb.init(
    project="exercise_5",
    job_type="eda",
    save_code=True
)
```

```python
artifact = run.use_artifact("exercise_4/genres_mod.parquet:latest")
df = pd.read_parquet(artifact.file())
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
df['title'].fillna(value='', inplace=True)
df['song_name'].fillna(value='', inplace=True)
df['text_feature'] = df['title'] + ' ' + df['song_name']
```

```python
df.to_csv("preprocessed_data.csv", index=False)
```

```python
preprocessed_artifact = wandb.Artifact(
    name="preprocessed_data.csv",
    type="dataset"
)
```

```python
preprocessed_artifact.add_file("preprocessed_data.csv")
run.log_artifact(preprocessed_artifact)
```

```python
run.finish()
```
